Создайте программу, демонстрирующую принципы полиморфизма,
перегрузки и перекрытия методов в контексте класса "Компьютер" и
его компонентов. Используйте обобщенные типы (Generic) для создания
гибкой системы.

```
using System;
using System.Collections.Generic;

// Интерфейс установки
public interface IInstallable
{
    void Install();
}

// Базовый абстрактный класс комплектующих (полиморфизм)
public abstract class Component
{
    public string Name { get; set; }
    public decimal Price { get; set; }

    public Component(string name, decimal price)
    {
        Name = name;
        Price = price;
    }

    // (полиморфизм)
    public virtual string GetInfo()
    {
        return $"{Name} | Стоимость: {Price} руб.";
    }

    // (перегрузка)
    public override string ToString()
    {
        return GetInfo();
    }
}

// Процессор
public class CPU : Component, IInstallable
{
    public int Cores { get; set; }

    public CPU(string name, decimal price, int cores)
        : base(name, price)
    {
        Cores = cores;
    }

    public override string GetInfo()
    {
        return $"Процессор: {Name} | Ядра: {Cores} | Цена: {Price} руб.";
    }

    public void Install()
    {
        Console.WriteLine($"Процессор {Name} установлен.");
    }

    public static bool operator ==(CPU a, CPU b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name && a.Price == b.Price && a.Cores == b.Cores;
    }

    public static bool operator !=(CPU a, CPU b) => !(a == b);

    public override bool Equals(object obj)
    {
        return obj is CPU cpu && this == cpu;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Price, Cores);
    }
}

// Оперативная память
public class RAM : Component, IInstallable
{
    public int Capacity { get; set; }

    public RAM(string name, decimal price, int capacity)
        : base(name, price)
    {
        Capacity = capacity;
    }

    public override string GetInfo()
    {
        return $"Оперативная память: {Name} | {Capacity} ГБ | Цена: {Price} руб.";
    }

    public void Install()
    {
        Console.WriteLine($"Модуль памяти {Name} установлен.");
    }

    public static bool operator ==(RAM a, RAM b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name && a.Price == b.Price && a.Capacity == b.Capacity;
    }

    public static bool operator !=(RAM a, RAM b) => !(a == b);

    public override bool Equals(object obj)
    {
        return obj is RAM ram && this == ram;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Price, Capacity);
    }
}

// Видеокарта
public class GPU : Component, IInstallable
{
    public int VideoMemory { get; set; }

    public GPU(string name, decimal price, int videoMemory)
        : base(name, price)
    {
        VideoMemory = videoMemory;
    }

    public override string GetInfo()
    {
        return $"Видеокарта: {Name} | Видеопамять: {VideoMemory} ГБ | Цена: {Price} руб.";
    }

    public void Install()
    {
        Console.WriteLine($"Видеокарта {Name} подключена.");
    }

    public static bool operator ==(GPU a, GPU b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name && a.Price == b.Price && a.VideoMemory == b.VideoMemory;
    }

    public static bool operator !=(GPU a, GPU b) => !(a == b);

    public override bool Equals(object obj)
    {
        return obj is GPU gpu && this == gpu;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Price, VideoMemory);
    }
}

// Накопитель
public class HDD : Component, IInstallable
{
    public int Capacity { get; set; }

    public HDD(string name, decimal price, int capacity)
        : base(name, price)
    {
        Capacity = capacity;
    }

    public override string GetInfo()
    {
        return $"Накопитель: {Name} | Объём: {Capacity} ГБ | Цена: {Price} руб.";
    }

    public void Install()
    {
        Console.WriteLine($"Накопитель {Name} подключён.");
    }

    public static bool operator ==(HDD a, HDD b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name && a.Price == b.Price && a.Capacity == b.Capacity;
    }

    public static bool operator !=(HDD a, HDD b) => !(a == b);

    public override bool Equals(object obj)
    {
        return obj is HDD hdd && this == hdd;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Price, Capacity);
    }
}

// Компьютер
public class Computer
{
    private List<Component> parts = new List<Component>();

    // Generic-метод добавления
    public void AddComponent<T>(T component)
        where T : Component, IInstallable
    {
        if (component == null)
        {
            Console.WriteLine("Нельзя добавить пустой компонент.");
            return;
        }

        parts.Add(component);
        component.Install();
    }

    // Generic-метод удаления
    public void RemoveComponent<T>(T component)
        where T : Component
    {
        if (parts.Remove(component))
        {
            Console.WriteLine($"Успешно удалено из ПК: {component.GetInfo()}");
        }
        else
        {
            Console.WriteLine($"\n[УДАЛЕНИЕ] Компонент '{component.Name}' не найден в данном компьютере.");
        }
    }

    // Подсчёт общей цены
    public decimal GetTotalPrice()
    {
        decimal total = 0;

        foreach (Component part in parts)
            total += part.Price;

        return total;
    }

    // Вывод характеристик
    public void PrintInfo()
    {
        Console.WriteLine("\n        КОМПЛЕКТАЦИЯ ПК          ");

        foreach (var c in parts)
        {
            // перекрытие
            Console.WriteLine("- " + c.ToString());
        }

        Console.WriteLine("-------------------------------------");
        Console.WriteLine($"Количество деталей: {parts.Count}");
        Console.WriteLine($"ИТОГО: {GetTotalPrice()} руб.");
        Console.WriteLine(" ");
    }
}

// Фабрика готовых компьютеров
public static class ComputerFactory
{
    public static Computer CreateStudyPC()
    {
        Computer pc = new Computer();

        pc.AddComponent(new CPU("AMD Ryzen 3 4100", 6500m, 4));
        pc.AddComponent(new RAM("Crucial DDR4", 2200m, 8));
        pc.AddComponent(new HDD("Kingston SSD", 3000m, 480));

        return pc;
    }

    public static Computer CreatePowerfulPC()
    {
        Computer pc = new Computer();

        pc.AddComponent(new CPU("AMD Ryzen 7 5700X", 17000m, 8));
        pc.AddComponent(new RAM("Corsair Vengeance", 6500m, 32));
        pc.AddComponent(new GPU("Radeon RX 6700 XT", 35000m, 12));
        pc.AddComponent(new HDD("Samsung SSD", 5500m, 1000));

        return pc;
    }
}

// ЗАПУСК ПРОГРАММЫ

Console.WriteLine("       КОНСТРУКТОР ОФИСНОГО КОМПЬЮТЕРА         ");
Computer studyPC = ComputerFactory.CreateStudyPC();
studyPC.PrintInfo();

Console.WriteLine("        ТЕСТ УДАЛЕНИЯ КОМПОНЕНТА        ");
var ramToRemove = new RAM("Crucial DDR4", 2200m, 8);
studyPC.RemoveComponent(ramToRemove);
studyPC.PrintInfo(); 

Console.WriteLine("       КОНСТРУКТОР ИГРОВОГО КОМПЬЮТЕРА        ");
Computer powerfulPC = ComputerFactory.CreatePowerfulPC();
powerfulPC.PrintInfo();

Console.WriteLine("        СРАВНЕНИЕ ПРОЦЕССОРОВ        ");

CPU firstCPU = new CPU("AMD Ryzen 3 4100", 6500m, 4);
CPU secondCPU = new CPU("AMD Ryzen 3 4100", 6500m, 4);
CPU thirdCPU = new CPU("AMD Ryzen 7 5700X", 17000m, 8);

Console.WriteLine($"Процессор 1: {firstCPU.GetInfo()}");
Console.WriteLine($"Процессор 2: {secondCPU.GetInfo()}");
Console.WriteLine($"Процессор 3: {thirdCPU.GetInfo()}");
Console.WriteLine("--------------------------------------------------");

Console.WriteLine($"Результат (Процессор 1 == Процессор 2): {firstCPU == secondCPU}");
Console.WriteLine($"Результат (Процессор 1 != Процессор 3): {firstCPU != thirdCPU}");
```
