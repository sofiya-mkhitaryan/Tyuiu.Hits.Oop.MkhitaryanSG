```
using System;
using System.Collections.Generic;

// Интерфейс для онлайн-мероприятий
public interface IOnlineEvent
{
    string Platform { get; set; }

    void ConnectOnline();
}

// Интерфейс для записи мероприятия
public interface IRecordable
{
    bool IsRecorded { get; set; }

    void StartRecording();
}


// Базовый класс
public class Event
{
    // Основные свойства
    public int EventId { get; set; }
    public string EventName { get; set; }
    public DateTime Date { get; set; }

    // Дополнительные свойства
    public string Location { get; set; }
    public int MaxParticipants { get; set; }

    protected List<string> Participants { get; set; } = new List<string>();

    // Конструктор
    public Event(
        int eventId,
        string eventName,
        DateTime date,
        string location,
        int maxParticipants)
    {
        EventId = eventId;
        EventName = eventName;
        Date = date;
        Location = location;
        MaxParticipants = maxParticipants;
    }

    // Метод регистрации участника
    public virtual void RegisterParticipant(string participantName)
    {
        if (Participants.Count < MaxParticipants)
        {
            Participants.Add(participantName);
            Console.WriteLine(
                $"Участник {participantName} зарегистрирован на мероприятие «{EventName}».");
        }
        else
        {
            Console.WriteLine("Достигнуто максимальное количество участников.");
        }
    }

    // Метод уведомления
    public virtual void NotifyParticipants()
    {
        Console.WriteLine(
            $"Уведомление: мероприятие «{EventName}» состоится {Date:dd.MM.yyyy}.");
    }

    // Метод вывода информации
    public virtual void GetEventDetails()
    {
        Console.WriteLine($"ID мероприятия: {EventId}");
        Console.WriteLine($"Название: {EventName}");
        Console.WriteLine($"Дата: {Date:dd.MM.yyyy}");
        Console.WriteLine($"Место проведения: {Location}");
        Console.WriteLine($"Максимальное количество участников: {MaxParticipants}");
        Console.WriteLine($"Зарегистрировано участников: {Participants.Count}");
    }

    // Дополнительный метод
    public void CancelEvent()
    {
        Console.WriteLine($"Мероприятие «{EventName}» отменено.");
    }

    // Взаимодействие объектов
    public void ShareParticipant(Event otherEvent, string participantName)
    {
        Console.WriteLine(
            $"--- Взаимодействие: «{EventName}» и «{otherEvent.EventName}» ---");

        RegisterParticipant(participantName);
        otherEvent.RegisterParticipant(participantName);
    }
}


// Производный класс Conference
// ПРОСТОЕ НАСЛЕДОВАНИЕ: Conference -> Event
public class Conference : Event
{
    public int SpeakersCount { get; set; }
    public string Program { get; set; }

    public Conference(
        int eventId,
        string eventName,
        DateTime date,
        string location,
        int maxParticipants,
        int speakersCount,
        string program)
        : base(eventId, eventName, date, location, maxParticipants)
    {
        SpeakersCount = speakersCount;
        Program = program;
    }

    public override void NotifyParticipants()
    {
        Console.WriteLine(
            $"Уведомление о конференции «{EventName}».");
        Console.WriteLine($"Дата: {Date:dd.MM.yyyy}");
        Console.WriteLine($"Количество спикеров: {SpeakersCount}");
    }

    public override void GetEventDetails()
    {
        base.GetEventDetails();
        Console.WriteLine($"Количество спикеров: {SpeakersCount}");
        Console.WriteLine($"Программа: {Program}");
    }

    public void StartConference()
    {
        Console.WriteLine($"Конференция «{EventName}» началась.");
    }
}


// Производный класс Festival
// ПРОСТОЕ НАСЛЕДОВАНИЕ: Festival -> Event
public class Festival : Event
{
    public int BandsCount { get; set; }
    public List<string> SelectedGenres { get; set; } =
        new List<string>();

    public Festival(
        int eventId,
        string eventName,
        DateTime date,
        string location,
        int maxParticipants,
        int bandsCount)
        : base(eventId, eventName, date, location, maxParticipants)
    {
        BandsCount = bandsCount;
    }

    public override void RegisterParticipant(string participantName)
    {
        Participants.Add(participantName);

        string genre = "Рок";
        SelectedGenres.Add(genre);

        Console.WriteLine(
            $"Участник {participantName} зарегистрирован на фестиваль «{EventName}».");
        Console.WriteLine($"Выбранный жанр: {genre}");
    }

    public void ShowFestivalProgram()
    {
        Console.WriteLine($"Фестиваль «{EventName}»");
        Console.WriteLine($"Количество групп: {BandsCount}");
        Console.WriteLine("Музыкальные жанры:");

        foreach (string genre in SelectedGenres)
        {
            Console.WriteLine($"- {genre}");
        }
    }

    public void AddGenre(string genre)
    {
        SelectedGenres.Add(genre);
        Console.WriteLine($"Жанр «{genre}» добавлен.");
    }
}


// Производный класс Seminar
// ПРОСТОЕ НАСЛЕДОВАНИЕ: Seminar -> Event
public class Seminar : Event
{
    public List<string> Topics { get; set; } =
        new List<string>();

    public string Lecturer { get; set; }

    public Seminar(
        int eventId,
        string eventName,
        DateTime date,
        string location,
        int maxParticipants,
        List<string> topics,
        string lecturer)
        : base(eventId, eventName, date, location, maxParticipants)
    {
        Topics = topics;
        Lecturer = lecturer;
    }

    public override void GetEventDetails()
    {
        base.GetEventDetails();

        Console.WriteLine($"Преподаватель: {Lecturer}");
        Console.WriteLine("Тематические области:");

        foreach (string topic in Topics)
        {
            Console.WriteLine($"- {topic}");
        }
    }

    public void AddTopic(string topic)
    {
        Topics.Add(topic);
        Console.WriteLine($"Тематическая область «{topic}» добавлена.");
    }

    public void StartSeminar()
    {
        Console.WriteLine($"Семинар «{EventName}» начался.");
    }
}


// СЛОЖНОЕ / МНОГОУРОВНЕВОЕ НАСЛЕДОВАНИЕ
// Event -> Conference -> OnlineConference
public class OnlineConference : Conference, IOnlineEvent
{
    public string Platform { get; set; }

    public OnlineConference(
        int eventId,
        string eventName,
        DateTime date,
        string location,
        int maxParticipants,
        int speakersCount,
        string program,
        string platform)
        : base(
            eventId,
            eventName,
            date,
            location,
            maxParticipants,
            speakersCount,
            program)
    {
        Platform = platform;
    }

    public void ConnectOnline()
    {
        Console.WriteLine(
            $"Подключение к конференции через платформу {Platform}.");
    }

    public override void GetEventDetails()
    {
        base.GetEventDetails();
        Console.WriteLine($"Онлайн-платформа: {Platform}");
    }
}


// МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ ЧЕРЕЗ ИНТЕРФЕЙСЫ
// HybridEvent наследуется от Event
// и одновременно реализует IOnlineEvent и IRecordable
public class HybridEvent : Event, IOnlineEvent, IRecordable
{
    public string Platform { get; set; }
    public bool IsRecorded { get; set; }

    public HybridEvent(
        int eventId,
        string eventName,
        DateTime date,
        string location,
        int maxParticipants,
        string platform)
        : base(
            eventId,
            eventName,
            date,
            location,
            maxParticipants)
    {
        Platform = platform;
        IsRecorded = false;
    }

    public void ConnectOnline()
    {
        Console.WriteLine(
            $"Онлайн-подключение к мероприятию через {Platform}.");
    }

    public void StartRecording()
    {
        IsRecorded = true;
        Console.WriteLine(
            $"Запись мероприятия «{EventName}» начата.");
    }

    public override void GetEventDetails()
    {
        base.GetEventDetails();
        Console.WriteLine($"Онлайн-платформа: {Platform}");
        Console.WriteLine($"Запись ведется: {IsRecorded}");
    }
}


// СОЗДАНИЕ ОБЪЕКТОВ

Conference conference = new Conference(
    1,
    "Международная IT-конференция",
    new DateTime(2026, 10, 15),
    "Москва",
    500,
    12,
    "Открытие; доклады; мастер-классы");

Festival festival = new Festival(
    2,
    "Музыкальный фестиваль",
    new DateTime(2026, 11, 5),
    "Санкт-Петербург",
    1000,
    15);

Seminar seminar = new Seminar(
    3,
    "Современные технологии программирования",
    new DateTime(2026, 10, 20),
    "Казань",
    100,
    new List<string>
    {
        "C# и .NET",
        "ООП",
        "Искусственный интеллект"
    },
    "Иван Петров");

OnlineConference onlineConference = new OnlineConference(
    4,
    "Онлайн-конференция по C#",
    new DateTime(2026, 12, 10),
    "Онлайн",
    1000,
    8,
    "C#; .NET; ASP.NET",
    "Zoom");

HybridEvent hybridEvent = new HybridEvent(
    5,
    "Гибридная IT-конференция",
    new DateTime(2026, 12, 20),
    "Москва",
    700,
    "Microsoft Teams");


// ПРОВЕРКА ПРОСТОГО НАСЛЕДОВАНИЯ

conference.RegisterParticipant("Иван Иванов");
conference.NotifyParticipants();
conference.GetEventDetails();
conference.StartConference();

Console.WriteLine();

festival.RegisterParticipant("Анна Петрова");
festival.AddGenre("Поп");
festival.ShowFestivalProgram();

Console.WriteLine();

seminar.RegisterParticipant("Пётр Сидоров");
seminar.AddTopic("Машинное обучение");
seminar.StartSeminar();


// ПРОВЕРКА СЛОЖНОГО НАСЛЕДОВАНИЯ

Console.WriteLine();
Console.WriteLine("=== Многоуровневое наследование ===");

onlineConference.RegisterParticipant("Мария Смирнова");
onlineConference.ConnectOnline();
onlineConference.GetEventDetails();


// ПРОВЕРКА МНОЖЕСТВЕННОГО НАСЛЕДОВАНИЯ

Console.WriteLine();
Console.WriteLine("=== Множественное наследование через интерфейсы ===");

hybridEvent.RegisterParticipant("Алексей Иванов");
hybridEvent.ConnectOnline();
hybridEvent.StartRecording();
hybridEvent.GetEventDetails();


// ВЗАИМОДЕЙСТВИЕ ОБЪЕКТОВ

Console.WriteLine();
Console.WriteLine("=== Взаимодействие объектов ===");

conference.ShareParticipant(
    seminar,
    "Общий участник");


// Работа со списком объектов базового типа

Console.WriteLine();
Console.WriteLine("=== Все мероприятия ===");

List<Event> events = new List<Event>
{
    conference,
    festival,
    seminar,
    onlineConference,
    hybridEvent
};

foreach (Event currentEvent in events)
{
    Console.WriteLine();
    currentEvent.GetEventDetails();
}
```
