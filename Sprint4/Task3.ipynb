Создать базовый класс Event в C#, который будет представлять организованные мероприятия. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.
Требования к базовому классу Event:

Атрибуты: ID мероприятия (EventId), Название мероприятия (EventName), Дата проведения (Date).

Методы:
  - RegisterParticipant(): метод для регистрации участника.
  - NotifyParticipants(): метод для уведомления участников о мероприятии.
  - GetEventDetails(): метод для получения деталей мероприятия.

Требования к производным классам:

1. Конференция (Conference): Должна содержать дополнительные атрибуты, такие как Количество спикеров (SpeakersCount). Метод NotifyParticipants() должен быть переопределен для отправки уведомлений с программой конференции.

2. Фестиваль (Festival): Должна содержать дополнительные атрибуты, такие как Количество музыкальных групп (BandsCount). Метод RegisterParticipant() должен быть переопределен для регистрации участников с выбором интересующих музыкальных жанров.

3. Семинар (Seminar) (если требуется третий класс): Должна содержать дополнительные атрибуты, такие как Тематические области (Topics). Метод GetEventDetails() должен быть переопределен для отображения тематических областей семинара вместе с другими деталями мероприятия.

```

using System;
using System.Collections.Generic;

// Интерфейсы
public interface IOnlineEvent
{
    string Platform { get; set; }
    void ConnectOnline();
}

public interface IRecordable
{
    bool IsRecorded { get; set; }
    void StartRecording();
}

public interface IEventActions
{
    void ShowAdditionalInfo();
}

// Базовый класс
public class Event : IEventActions
{
    public int EventId { get; set; }
    public string EventName { get; set; }
    public DateTime Date { get; set; }
    public string Location { get; set; }
    public int MaxParticipants { get; set; }
    public decimal TicketPrice { get; set; }

    public string Organizer { get; set; }
    public string Category { get; set; }
    public bool IsActive { get; set; }

    protected List<string> Participants { get; set; } = new List<string>();

    public Event(int eventId, string eventName, DateTime date,
        string location, int maxParticipants, decimal ticketPrice)
    {
        EventId = eventId;
        EventName = eventName;
        Date = date;
        Location = location;
        MaxParticipants = maxParticipants;
        TicketPrice = ticketPrice;
        Organizer = "Не указан";
        Category = "Общее";
        IsActive = true;
    }

    public virtual void RegisterParticipant(string participantName)
    {
        if (Participants.Count < MaxParticipants)
        {
            Participants.Add(participantName);
            Console.WriteLine($"Участник {participantName} зарегистрирован на «{EventName}».");
        }
        else
        {
            Console.WriteLine("Достигнуто максимальное количество участников.");
        }
    }

    public void RegisterParticipant(string participantName, string email)
    {
        RegisterParticipant(participantName);
        Console.WriteLine($"Email: {email}");
    }

    public virtual void NotifyParticipants()
    {
        Console.WriteLine($"Мероприятие «{EventName}» состоится {Date:dd.MM.yyyy}.");
    }

    public virtual void GetEventDetails()
    {
        Console.WriteLine($"ID: {EventId}");
        Console.WriteLine($"Название: {EventName}");
        Console.WriteLine($"Дата: {Date:dd.MM.yyyy}");
        Console.WriteLine($"Место: {Location}");
        Console.WriteLine($"Максимум участников: {MaxParticipants}");
        Console.WriteLine($"Стоимость билета: {TicketPrice} руб.");
        Console.WriteLine($"Организатор: {Organizer}");
        Console.WriteLine($"Категория: {Category}");
        Console.WriteLine($"Активно: {IsActive}");
        Console.WriteLine($"Участников: {Participants.Count}");
    }

    public void CancelEvent()
    {
        IsActive = false;
        Console.WriteLine($"Мероприятие «{EventName}» отменено.");
    }

    public void ChangeLocation(string newLocation)
    {
        Location = newLocation;
        Console.WriteLine($"Место проведения изменено: {Location}");
    }

    public void ChangeTicketPrice(decimal newPrice)
    {
        TicketPrice = newPrice;
        Console.WriteLine($"Новая стоимость билета: {TicketPrice} руб.");
    }

    public void ShareParticipant(Event otherEvent, string participantName)
    {
        RegisterParticipant(participantName);
        otherEvent.RegisterParticipant(participantName);
    }

    void IEventActions.ShowAdditionalInfo()
    {
        Console.WriteLine($"Дополнительная информация: организатор — {Organizer}, категория — {Category}.");
    }
}

// Конференция
public class Conference : Event, IEventActions
{
    public int SpeakersCount { get; set; }
    public string Program { get; set; }
    public string SponsorName { get; set; }

    public string MainTopic { get; set; }
    public bool HasWorkshop { get; set; }

    public Conference(int eventId, string eventName, DateTime date,
        string location, int maxParticipants, decimal ticketPrice,
        int speakersCount, string program, string sponsorName)
        : base(eventId, eventName, date, location, maxParticipants, ticketPrice)
    {
        SpeakersCount = speakersCount;
        Program = program;
        SponsorName = sponsorName;
        MainTopic = "Не указана";
        HasWorkshop = false;
    }

    public override void NotifyParticipants()
    {
        Console.WriteLine($"Уведомление о конференции «{EventName}».");
        Console.WriteLine($"Дата: {Date:dd.MM.yyyy}");
        Console.WriteLine($"Спикеров: {SpeakersCount}");
    }

    public override void GetEventDetails()
    {
        base.GetEventDetails();
        Console.WriteLine($"Спикеров: {SpeakersCount}");
        Console.WriteLine($"Программа: {Program}");
        Console.WriteLine($"Спонсор: {SponsorName}");
        Console.WriteLine($"Главная тема: {MainTopic}");
        Console.WriteLine($"Есть мастер-класс: {HasWorkshop}");
    }

    public void StartConference()
    {
        Console.WriteLine($"Конференция «{EventName}» началась.");
    }

    public void AddSpeaker()
    {
        SpeakersCount++;
        Console.WriteLine($"Количество спикеров: {SpeakersCount}");
    }

    void IEventActions.ShowAdditionalInfo()
    {
        Console.WriteLine($"Конференция: {MainTopic}. Спонсор: {SponsorName}.");
    }
}

// Фестиваль
public class Festival : Event, IEventActions
{
    public int BandsCount { get; set; }
    public List<string> SelectedGenres { get; set; } = new List<string>();
    public bool IsOpenAir { get; set; }

    public string Headliner { get; set; }
    public int DurationDays { get; set; }

    public Festival(int eventId, string eventName, DateTime date,
        string location, int maxParticipants, decimal ticketPrice,
        int bandsCount, bool isOpenAir)
        : base(eventId, eventName, date, location, maxParticipants, ticketPrice)
    {
        BandsCount = bandsCount;
        IsOpenAir = isOpenAir;
        Headliner = "Не указан";
        DurationDays = 1;
    }

    public override void RegisterParticipant(string participantName)
    {
        if (Participants.Count < MaxParticipants)
        {
            Participants.Add(participantName);
            Console.WriteLine($"Участник {participantName} зарегистрирован на фестиваль.");
        }
        else
        {
            Console.WriteLine("Достигнуто максимальное количество участников.");
        }
    }

    public void ShowFestivalProgram()
    {
        Console.WriteLine($"Фестиваль «{EventName}»");
        Console.WriteLine($"Групп: {BandsCount}");
        Console.WriteLine($"На открытом воздухе: {(IsOpenAir ? "Да" : "Нет")}");
        Console.WriteLine($"Главный исполнитель: {Headliner}");
        Console.WriteLine($"Продолжительность: {DurationDays} дней");

        foreach (string genre in SelectedGenres)
            Console.WriteLine($"- {genre}");
    }

    public void AddGenre(string genre)
    {
        SelectedGenres.Add(genre);
        Console.WriteLine($"Жанр «{genre}» добавлен.");
    }

    public void AddBand()
    {
        BandsCount++;
        Console.WriteLine($"Количество групп: {BandsCount}");
    }

    void IEventActions.ShowAdditionalInfo()
    {
        Console.WriteLine($"Фестиваль: главный исполнитель — {Headliner}, длительность — {DurationDays} дней.");
    }
}

// Семинар
public class Seminar : Event, IEventActions
{
    public List<string> Topics { get; set; } = new List<string>();
    public string Lecturer { get; set; }
    public bool ProvideCertificate { get; set; }

    public string DifficultyLevel { get; set; }
    public int DurationHours { get; set; }

    public Seminar(int eventId, string eventName, DateTime date,
        string location, int maxParticipants, decimal ticketPrice,
        List<string> topics, string lecturer, bool provideCertificate)
        : base(eventId, eventName, date, location, maxParticipants, ticketPrice)
    {
        Topics = topics;
        Lecturer = lecturer;
        ProvideCertificate = provideCertificate;
        DifficultyLevel = "Начальный";
        DurationHours = 2;
    }

    public override void GetEventDetails()
    {
        base.GetEventDetails();
        Console.WriteLine($"Преподаватель: {Lecturer}");
        Console.WriteLine($"Сертификат: {(ProvideCertificate ? "Да" : "Нет")}");
        Console.WriteLine($"Уровень сложности: {DifficultyLevel}");
        Console.WriteLine($"Продолжительность: {DurationHours} часов");

        foreach (string topic in Topics)
            Console.WriteLine($"- {topic}");
    }

    public void AddTopic(string topic)
    {
        Topics.Add(topic);
        Console.WriteLine($"Тема «{topic}» добавлена.");
    }

    public void StartSeminar()
    {
        Console.WriteLine($"Семинар «{EventName}» начался.");
    }

    public void ExtendSeminar(int hours)
    {
        DurationHours += hours;
        Console.WriteLine($"Новая продолжительность: {DurationHours} часов.");
    }

    void IEventActions.ShowAdditionalInfo()
    {
        Console.WriteLine($"Семинар ведёт {Lecturer}. Уровень сложности: {DifficultyLevel}.");
    }
}

// Многоуровневое наследование
public class OnlineConference : Conference, IOnlineEvent
{
    public string Platform { get; set; }

    public OnlineConference(int eventId, string eventName, DateTime date,
        string location, int maxParticipants, decimal ticketPrice,
        int speakersCount, string program, string sponsorName, string platform)
        : base(eventId, eventName, date, location, maxParticipants,
            ticketPrice, speakersCount, program, sponsorName)
    {
        Platform = platform;
    }

    public void ConnectOnline()
    {
        Console.WriteLine($"Подключение через {Platform}.");
    }

    public override void GetEventDetails()
    {
        base.GetEventDetails();
        Console.WriteLine($"Онлайн-платформа: {Platform}");
    }
}

// Множественное наследование через интерфейсы
public class HybridEvent : Event, IOnlineEvent, IRecordable
{
    public string Platform { get; set; }
    public bool IsRecorded { get; set; }

    public HybridEvent(int eventId, string eventName, DateTime date,
        string location, int maxParticipants, decimal ticketPrice, string platform)
        : base(eventId, eventName, date, location, maxParticipants, ticketPrice)
    {
        Platform = platform;
        IsRecorded = false;
    }

    public void ConnectOnline()
    {
        Console.WriteLine($"Подключение через {Platform}.");
    }

    public void StartRecording()
    {
        IsRecorded = true;
        Console.WriteLine($"Запись мероприятия «{EventName}» начата.");
    }

    public override void GetEventDetails()
    {
        base.GetEventDetails();
        Console.WriteLine($"Платформа: {Platform}");
        Console.WriteLine($"Запись ведется: {IsRecorded}");
    }
}

// Управление зависимостями
public class EventService
{
    private readonly IEventActions _eventActions;

    public EventService(IEventActions eventActions)
    {
        _eventActions = eventActions;
    }

    public void ShowEventInformation()
    {
        _eventActions.ShowAdditionalInfo();
    }
}

// Генерический класс
public class EventCollection<T> where T : Event
{
    private List<T> _events = new List<T>();

    public void Add(T eventItem)
    {
        _events.Add(eventItem);
    }

    public void DisplayAllDetails()
    {
        foreach (T item in _events)
        {
            item.GetEventDetails();
            Console.WriteLine(new string('-', 30));
        }
    }
}

// Создание объектов
Conference conference = new Conference(
    1, "IT-конференция", new DateTime(2026, 10, 15),
    "Москва", 500, 5000m, 12,
    "Доклады и мастер-классы", "ТехноСофт");

Festival festival = new Festival(
    2, "Музыкальный фестиваль", new DateTime(2026, 11, 5),
    "Санкт-Петербург", 1000, 1500m, 15, true);

Seminar seminar = new Seminar(
    3, "Программирование на C#", new DateTime(2026, 10, 20),
    "Казань", 100, 2000m,
    new List<string> { "C#", "ООП", ".NET" },
    "Иван Петров", true);

OnlineConference onlineConference = new OnlineConference(
    4, "Онлайн-конференция", new DateTime(2026, 12, 10),
    "Онлайн", 1000, 1000m, 8,
    "C# и .NET", "IT-Академия", "Zoom");

HybridEvent hybridEvent = new HybridEvent(
    5, "Гибридная конференция", new DateTime(2026, 12, 20),
    "Москва", 700, 3000m, "Microsoft Teams");

// Проверка новых свойств и методов
Console.WriteLine("=== Новые возможности ===");

conference.Organizer = "IT Academy";
conference.Category = "Образование";
conference.MainTopic = "Разработка ПО";
conference.HasWorkshop = true;
conference.AddSpeaker();
conference.ChangeTicketPrice(4500m);
conference.GetEventDetails();

Console.WriteLine();

festival.Headliner = "Группа Example";
festival.DurationDays = 3;
festival.AddGenre("Рок");
festival.AddBand();
festival.ShowFestivalProgram();

Console.WriteLine();

seminar.DifficultyLevel = "Продвинутый";
seminar.ExtendSeminar(2);
seminar.AddTopic("Искусственный интеллект");
seminar.GetEventDetails();

// Явная реализация интерфейса
Console.WriteLine("\n=== Явная реализация интерфейса ===");

IEventActions conferenceActions = conference;
conferenceActions.ShowAdditionalInfo();

IEventActions festivalActions = festival;
festivalActions.ShowAdditionalInfo();

IEventActions seminarActions = seminar;
seminarActions.ShowAdditionalInfo();

// Управление зависимостями
Console.WriteLine("\n=== Dependency Injection ===");

EventService conferenceService = new EventService(conference);
conferenceService.ShowEventInformation();

EventService festivalService = new EventService(festival);
festivalService.ShowEventInformation();

EventService seminarService = new EventService(seminar);
seminarService.ShowEventInformation();

// Проверка онлайн и гибридных мероприятий
Console.WriteLine("\n=== Онлайн-мероприятия ===");

onlineConference.ConnectOnline();

hybridEvent.ConnectOnline();
hybridEvent.StartRecording();
hybridEvent.GetEventDetails();

// Взаимодействие объектов
Console.WriteLine("\n=== Взаимодействие объектов ===");

conference.ShareParticipant(seminar, "Иван Иванов");

// Генерический класс
Console.WriteLine("\n=== Коллекция мероприятий ===");

EventCollection<Event> myEvents = new EventCollection<Event>();

myEvents.Add(conference);
myEvents.Add(festival);
myEvents.Add(seminar);
myEvents.Add(hybridEvent);

myEvents.DisplayAllDetails();
```
