Создайте проект на C#, реализующее классы Товар (Product), Продавец (Seller), Магазин (Store) с использование геттеров и сеттеров, а также модификаторов доступа public, private, protected, internal. От каждого класса создайте по одному экземпляру.


```
using System;


// 1. Информация о магазине
Store store = new Store("Супермаркет Овощи-Фрукты", "ул. Ленина, 10", "09:00–21:00");
Console.WriteLine(store.GetStoreInfo());

Console.WriteLine();

// 2. Информация о продавце
Seller seller = new Seller("Иванов И.И.", "E123", 50000m, "123-45-6789");
store.AddSeller(seller);
Console.WriteLine(seller.GetSellerInfo());

Console.WriteLine();

// 3. Информация о товаре
Product product = new Product("Яблоки", 150.5m, 10, "Фрукты", "Свежее яблоко, без червей");
seller.AddProduct(product);
Console.WriteLine(product.GetProductInfo());
Console.WriteLine("Цена до изменения: " + product.GetPrice());
product.SetPrice(200m); 
Console.WriteLine("Цена после изменения: " + product.GetPrice());
seller.SellProduct(product, 2);

public class Product
{
    private string name;        
    private decimal price;       
    private int quantity;       
    protected string category;   
    internal string description; 

    public Product(string name, decimal price, int quantity, string category, string description = "")
    {
        this.name = name;
        this.price = price;
        this.quantity = quantity;
        this.category = category;
        this.description = description;
    }

    public string GetName() => name;
    public void SetName(string value)
    {
        if (!string.IsNullOrEmpty(value)) name = value;
    }

    public decimal GetPrice() => price;
    public void SetPrice(decimal value)
    {
        if (value >= 0) price = value;
    }

    public int GetQuantity() => quantity;
    public void SetQuantity(int value)
    {
        if (value >= 0) quantity = value;
    }

    protected string GetCategory() => category;
    protected void SetCategory(string category) => this.category = category;

    internal string GetDescription() => description;
    internal void SetDescription(string value)
    {
        if (value != null) description = value;
    }

    public decimal GetTotalPrice() => price * quantity;

    public void UpdateQuantity(int amount)
    {
        quantity += amount;
        if (quantity < 0) quantity = 0;
    }

    public string GetProductInfo() =>
        $"Товар: {name}, Цена: {price} руб., Количество: {quantity} шт., Категория: {category}";
}

public class Seller
{
    private string name;         
    private string employeeId;    
    protected decimal salary;    
    internal string contactInfo;  

    public Seller(string name, string employeeId, decimal salary, string contactInfo)
    {
        this.name = name;
        this.employeeId = employeeId;
        this.salary = salary;
        this.contactInfo = contactInfo;
    }

    public string GetName() => name;
    public void SetName(string value)
    {
        if (!string.IsNullOrEmpty(value)) name = value;
    }

    protected decimal GetSalary() => salary;
    protected void SetSalary(decimal value)
    {
        if (value >= 0) salary = value;
    }

    internal string GetContactInfo() => contactInfo;
    internal void SetContactInfo(string value)
    {
        if (!string.IsNullOrEmpty(value)) contactInfo = value;
    }

    public void AddProduct(Product product)
    {
        Console.WriteLine($"{name} добавил товар: {product.GetName()}");
    }

    public void SellProduct(Product product, int quantity)
    {
        product.UpdateQuantity(-quantity);
        Console.WriteLine($"{name} продал {quantity} единиц товара. Остаток: {product.GetQuantity()}");
    }

    public string GetSellerInfo() =>
        $"Продавец: {name}, ID: {employeeId}, Зарплата: {salary} руб., Контакты: {contactInfo}";
}

public class Store
{
    private string storeName;        
    private string location;        
    public string storeHours;        

    public Store(string storeName, string location, string storeHours)
    {
        this.storeName = storeName;
        this.location = location;
        this.storeHours = storeHours;
    }

    public string GetName() => storeName;

    public void AddSeller(Seller seller)
    {
        Console.WriteLine($"{storeName} нанял продавца: {seller.GetName()}");
    }

    public void ListProducts()
    {
        Console.WriteLine($"Список товаров в магазине {storeName}:");
        Console.WriteLine("Товары за продавцами пока не закреплены в базе.");
    }

    public string GetStoreInfo() =>
        $"Магазин: {storeName}, Расположение: {location}, Часы работы: {storeHours}";
}
```
